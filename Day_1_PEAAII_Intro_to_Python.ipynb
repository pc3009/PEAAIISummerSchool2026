```python
# Object-Oriented Programming
# Class - data structure

# THIS IS A COMMENT

class Person:

  # Constructor function
  def __init__(self, name, age, hair_color):
    self.name = name
    self.age = age
    self.hair_color = hair_color

  def greet(self, stranger):
    print("Hello " + stranger + ", my name is " + self.name + " !")
```

```python
teacher = Person(name = "Inal", age = 25, hair_color = "black")
```

```python
teacher.greet(stranger = "Justin")
```

# Goal

Take an input integer $N$, for every number in the range from $0$ to $N$, do:

if the number is even => print **Fizz**


else if the number is odd => print **Buzz**


else if the number is divisible by 5 => **FizzBuzz**

```python
# the solution function:
def solution(N):
  for number in range(0, N+1):
    if number % 2 == 0 and number% 5 != 0:
      print(str(number) + " Fizz")
    elif number % 5 == 0:
      print(str(number) + " FizzBuzz")
    elif number % 2 == 1:
      print(str(number) + " Buzz")


```

```python
solution(N = 51)
```
